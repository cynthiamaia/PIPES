```python
arquivo_csv = '/SVC.csv'
df = pd.read_csv(arquivo_csv)
if 'Flow ID' in df.columns:    
    resultados = []   
    for flow_id in df['Flow ID']:
        try:           
            flow_details = openml.flows.get_flow(int(flow_id))
            flow_name = flow_details.name   
            runs = openml.runs.list_runs(flow=[int(flow_id)])
            num_runs = len(runs)            
            print(f"Flow ID {flow_id} ({flow_name}): {num_runs}")          
            resultados.append({'Flow ID': flow_id, 'Flow Name': flow_name, 'Number of Executions': num_runs})
        except Exception as e:
            print(f"Error Flow ID {flow_id}: {e}")
            resultados.append({'Flow ID': flow_id, 'Flow Name': 'Error', 'Number of Executions': 'Error'})
    resultados_df = pd.DataFrame(resultados)
    resultados_df.to_csv('/Users/moreira/Documents/Meta_v_2/analises_openml/flow_id_executions_svc.csv', index=False)
else:
    print("The 'Flow ID' column was not found in the CSV file.")
```

```python
file_path = '/flow_id_executions_minmax.csv' 

columns = ['Flow ID', 'Number of Executions', ' Datasets']
df = pd.read_csv(file_path, usecols=columns)


execution_sum = df.groupby(' Datasets')['Number of Executions'].sum()
execution_sum = execution_sum.sort_values(ascending=False)
total_executions = execution_sum.sum()
print(total_executions)


execution_percentage = (execution_sum / total_executions) * 100
result = pd.DataFrame({
    'Dataset': execution_sum.index,
    'Total Executions': execution_sum.values,
    'Percentage (%)': execution_percentage.values
})
print(result)

```
